```python
import os
import json
from collections import defaultdict


DIFF_NAMES = ["easy", "medium", "multi_easy", "multi_medium", "undecidable"]


def get_bucket(i, difficulties):
    """
    difficulties: [i0, i1, i2, i3]
    index 기준으로 구간 나눔
    """

    i0, i1, i2, i3 = difficulties

    if i < i0:
        return "easy"
    elif i < i1:
        return "medium"
    elif i < i2:
        return "multi_easy"
    elif i < i3:
        return "multi_medium"
    else:
        return "undecidable"


def evaluate_one(results, difficulties):

    acc_dict = defaultdict(list)
    vis_dict = defaultdict(list)

    all_acc = []
    all_vis = []

    for i, r in enumerate(results):

        bucket = get_bucket(i, difficulties)

        acc = r["correct"]
        vis = r.get("target_object_visibility_gen_over_gt", None)
        vis = vis if vis is not None else 0

        acc_dict[bucket].append(acc)
        vis_dict[bucket].append(vis)

        all_acc.append(acc)
        all_vis.append(vis)


    for name in DIFF_NAMES:

        n = len(acc_dict[name])

        if n == 0:
            print(f"{name:12s} | n=0")
            continue

        acc = sum(acc_dict[name] * 100) / n
        vis = sum(vis_dict[name] * 100) / n

        print(f"{name:12s} | n={n:4d} | Acc: {acc:.2f} | Vis: {vis:.2f}")
    
    if len(all_acc) > 0:

        acc = sum(all_acc) / len(all_acc)
        vis = sum(all_vis) / len(all_vis)

        print("---------------------------------------")
        print(f"{'overall':12s} | n={len(all_acc):4d} | Acc: {acc:.2f} | Vis: {vis:.2f}")


def load_latest(result_dir, subdir):

    latest = sorted(os.listdir(os.path.join(result_dir, subdir)))[-1]

    path = os.path.join(result_dir, subdir, latest, "results.json")

    with open(path, "r") as f:
        results = json.load(f)

    return results


def print_evaluation(result_dir):

    for name, difficulties in zip(["existence_eval", 
    #"counting_eval", 
    "state_eval"], [
        [150, 300, 400, 445],
        #[150, 300, 352, 359],
        [150, 300, 400, 445],
    ]):

        print("\n====================")
        print(name)
        print("====================")

        results = load_latest(result_dir, name)

        evaluate_one(results, difficulties)
```
